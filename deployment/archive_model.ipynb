# Archive Model
Rama **FALSE** del gate. Marca la versión con tag `status=archived` y quita el alias `challenger` si existía. Idempotente: re-ejecutar no cambia nada.

```python
import mlflow
from mlflow.tracking import MlflowClient
mlflow.set_registry_uri("databricks-uc")

MODEL_NAME = dbutils.widgets.get("model_name")
MODEL_VERSION = dbutils.widgets.get("model_version")

client = MlflowClient()
client.set_model_version_tag(MODEL_NAME, MODEL_VERSION, "status", "archived")
try:
    client.delete_registered_model_alias(MODEL_NAME, "challenger")
except Exception:
    pass
print(f"{MODEL_NAME} v{MODEL_VERSION} archivado (tag status=archived)")
```
